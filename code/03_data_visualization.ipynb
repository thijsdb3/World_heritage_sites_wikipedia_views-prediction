```python

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

df = pd.read_csv("../data/processed/final_heritage_jan_2026.csv")

# Keep valid pageview values
views = pd.to_numeric(df["total_views"], errors="coerce")
views = views[views >= 0].dropna()

plt.figure(figsize=(9, 5))

sns.histplot(
    np.log1p(views),
    bins=40
)

# resembles a normal curve
plt.xlabel("Log  pageviews")
plt.ylabel("World Heritage properties")
plt.title("Distribution of Wikipedia Pageviews, Jan 2026")
plt.tight_layout()
plt.show()



```

```python
plt.figure(figsize=(9, 5))

sns.histplot(
    views,
    bins=40
)

plt.xlabel("pageviews")
plt.ylabel("World Heritage properties")
plt.title("Distribution of Wikipedia Pageviews, Jan 2026")
plt.tight_layout()
plt.show()
```
